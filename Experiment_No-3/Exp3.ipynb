**Simple Linear Regression**

```python
import pandas as pd
df=pd.read_csv('/content/drive/MyDrive/ML/scores.csv')
df.head(20)
```

**Preprocessing**

```python
col=['Scores']
df[col]=(df[col]-df[col].min())/(df[col].max()-df[col].min())
df.head(20)
```

**Split Data**

```python
 #Define features (X) and target (y) for this dataset
# Assuming 'Hours' is the feature and 'Scores' is the target
X_scores = df[['Hours']]
y_scores = df['Scores']
X_train_scores, X_test_scores, y_train_scores, y_test_scores = train_test_split(X_scores, y_scores, test_size=0.2, random_state=42)
```

**Train Linear Regression Model**

```python
from sklearn.linear_model import LinearRegression

# Initialize the Linear Regression model
model_scores = LinearRegression()

# Train the model using the training data
model_scores.fit(X_train_scores, y_train_scores)
```

**Make Predictions**

```python
# Make predictions on the test set
y_pred_scores = model_scores.predict(X_test_scores)

print("Predictions on the test set:")
# Display the first few predictions alongside actual values for comparison
predictions_df = pd.DataFrame({'Actual': y_test_scores.squeeze(), 'Predicted': y_pred_scores})

display(predictions_df.head(20))
```

**Evaluate Model**

```python
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Calculate Mean Squared Error (MSE)
mse_scores = mean_squared_error(y_test_scores, y_pred_scores)

print(f"Mean Squared Error (MSE): {mse_scores:.4f}")
```

**Plot Regression Line**

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(x=X_test_scores['Hours'].squeeze(), y=y_test_scores, label='Actual Scores', s=100, alpha=0.7)
sns.lineplot(x=X_test_scores['Hours'].squeeze(), y=y_pred_scores, color='red', label='Regression Line', linewidth=2)
plt.title('Regression Line: Hours vs Scores (Test Set)')
plt.xlabel('Hours Studied')
plt.ylabel('Scores')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```
